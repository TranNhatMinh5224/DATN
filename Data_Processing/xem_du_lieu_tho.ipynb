```python
import pandas as pd
import psycopg2
from sqlalchemy import create_engine

# Thông tin kết nối PostgreSQL
DB_USER = 'postgres'
DB_PASS = '12122004'
DB_HOST = 'localhost'
DB_PORT = '5432'
DB_NAME = 'mimic_iv'

# Tạo engine kết nối
engine = create_engine(f'postgresql://{DB_USER}:{DB_PASS}@{DB_HOST}:{DB_PORT}/{DB_NAME}')

def view_table(schema, table_name, limit=30):
    query = f'SELECT * FROM {schema}.{table_name} LIMIT {limit}'
    df = pd.read_sql(query, engine)
    print(f'\n--- BẢNG {schema}.{table_name} (Hiển thị {limit} dòng) ---')
    display(df)
```

```python
# 1. Nhóm Đặc trưng Chung (Nhân khẩu học & Hành trình)
view_table('hosp', 'patients')
view_table('hosp', 'admissions')
```

```python
# 2. Nhóm Bệnh nền & Từ điển ICD (Phục vụ Private B)
view_table('hosp', 'diagnoses_icd')
view_table('hosp', 'd_icd_diagnoses')
```
